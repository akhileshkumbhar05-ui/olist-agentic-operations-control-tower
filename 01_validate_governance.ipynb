```python

verification = spark.sql("""
SELECT
    (SELECT COUNT(*)
     FROM workspace.olist_governance.dq_rules)
        AS total_rules,

    (SELECT COUNT(*)
     FROM workspace.olist_governance.source_dictionary)
        AS source_fields,

    (SELECT COUNT(*)
     FROM workspace.olist_governance.metric_dictionary)
        AS metrics,

    (SELECT COUNT(DISTINCT rule_id)
     FROM workspace.olist_governance.rule_metric_impact)
        AS rules_with_impact_metadata,

    (SELECT COUNT(*)
     FROM workspace.olist_governance.dq_rules
     WHERE plain_english_description IS NULL
        OR TRIM(plain_english_description) = '')
        AS rules_missing_explanations
""")

display(verification)
```
