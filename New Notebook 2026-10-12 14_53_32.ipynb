```python

import mlflow

mlflow.set_tracking_uri("databricks")

mlflow.set_experiment(
    "/Shared/olist-mlflow-trace-diagnostic"
)

@mlflow.trace(name="diagnostic.root")
def test_tracing():
    with mlflow.start_span(
        name="diagnostic.tool",
        span_type="TOOL"
    ) as span:
        span.set_attribute("test_type", "storage_check")
        return "Trace storage test successful"

print(test_tracing())
```

```python

import mlflow

mlflow.set_tracking_uri("databricks")
mlflow.set_experiment(
    experiment_id="2462690133412615"
)

@mlflow.trace(name="diagnostic.notebook_to_original")
def test_original_experiment():
    with mlflow.start_span(
        name="diagnostic.tool",
        span_type="TOOL"
    ) as span:
        span.set_attribute(
            "source", "serverless_notebook"
        )
        return "Original experiment storage test passed"

print(test_original_experiment())
```
